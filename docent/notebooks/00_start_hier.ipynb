## 2. Kernbegrippen in het Internet of Things (IoT)

Het **Internet of Things (IoT)** verwijst naar het netwerk van fysieke voorwerpen ("dingen") die uitgerust zijn met sensoren, software en draadloze technologie om gegevens te verzamelen en uit te wisselen via het internet.

Om de volgende onderdelen goed te begrijpen, vind je hieronder een overzicht van de belangrijkste vaktermen met hun exacte betekenis:

| Begrip | Wikipedia-stijl uitleg & Betekenis in deze les |
|---|---|
| **IoT-toestel / node / eindpunt** | Een fysiek apparaat met een micro-controller, sensor en radiozender dat zelfstandig metingen uitvoert en verstuurt (bijv. de *Dragino S31* klimaatsensor). |
| **Verbindingstype (Fysieke laag)** | Het draadloze of bedrade medium waarlangs digitale bits reizen: bijv. *Wi-Fi, LoRaWAN, Bluetooth, NB-IoT*. Vergelijk dit met het type weg (fietspad, snelweg, spoor). |
| **Gegevensuitwisselingsprotocol (Applicatielaag)** | De afgesproken regels en het formaat over hoe berichten gestructureerd zijn en uitgewisseld worden: bijv. *MQTT, HTTP, CoAP, Modbus*. Vergelijk dit met de taal en etiquette op de weg. |
| **Gateway** | Een tussenstation/netwerkapparaat dat draadloze radiosignalen (zoals LoRaWAN) ontvangt en vertaalt naar standaard internetverkeer (zoals Ethernet of Wi-Fi). |
| **Broker** | De centrale server bij het MQTT-protocol die berichten van zenders ontvangt en automatisch doorstuurt naar alle geïnteresseerde ontvangers. |
| **Topic** | Een gestructureerd digitaal adres (pad met schuine strepen, bijv. `stal1/afdeling3/temperatuur`) waarop berichten bij een broker gepubliceerd worden. |
| **Payload** | De eigenlijke inhoud van een bericht (de nuttige meetdata), los van alle netwerk- en adresinformatie. Dit kan bestaan uit ruwe getallen (bytes) of gestructureerde tekst (JSON). |
| **JSON (JavaScript Object Notation)** | Een lichtgewicht, menselijk leesbaar tekstformaat voor gegevensuitwisseling met sleutel-waarde-paren, bijv.: `{"temperatuur": 21.4, "vochtigheid": 63.2}`. |
| **Uplink / Downlink** | **Uplink** = bericht van de sensor naar de server/cloud. **Downlink** = bericht of commando van de server/cloud terug naar de sensor. |

**Rode draad in deze les:** Een temperatuur- en relatieve vochtigheidssensor in een stalafdeling voert elke 10 minuten een meting uit, verstuurt deze draadloos, waarna de data via een broker belandt op het scherm van een landbouwer of in de klimaatcomputer van de stal.

## 3. Verbinding met een MQTT-broker testen

In dit onderdeel testen we het **MQTT-protocol** (Message Queuing Telemetry Transport). MQTT is een extreem lichtgewicht en efficiënt protocol dat werkt volgens het **Publish-Subscribe (publiceren-abonneren)** principe:
1. Een sensor **publiceert** (*publish*) een bericht naar een specifiek **topic** op de broker.
2. Een ontvanger (bijv. een dashboard) **abonneert** (*subscribe*) zich op dat topic.
3. De **broker** ontvangt het bericht en stuurt het onmiddellijk door naar iedereen die geabonneerd is.

We maken hier gebruik van een publiek toegankelijke test-broker (`broker.hivemq.com`) op standaardpoort `1883`. Omdat deze broker openbaar is, genereren we een unieke willekeurige code (UUID) voor ons test-topic.

````python
import paho.mqtt.client as mqtt  # Importeer de MQTT-client module
import time                      # Importeer de tijdsmodule voor pauzes en time-outs
import uuid                      # Importeer de UUID-module om unieke tekstcodes te genereren

# Netwerk-instellingen voor de MQTT-broker
BROKER = "broker.hivemq.com"                            # Het adres van de centrale MQTT-server
POORT  = 1883                                            # Standaard netwerkpoort voor onversleuteld MQTT-verkeer
TOPIC  = f"ugent/i700301/test/{uuid.uuid4().hex[:8]}"   # Genereer een uniek topic-adres voor deze sessie

# Maak een lege lijst aan waarin we inkomende berichten opslaan
ontvangen = []

# Definieer een zogenaamde 'callback-functie': deze functie wordt automatisch door Python opgeroepen
# op het exacte moment dat er een nieuw bericht op het geabonneerde topic binnenkomt
def bij_bericht(client, userdata, msg):
    # msg.payload bevat de ruwe bytes; .decode() zet deze om naar leesbare tekst (string)
    ontvangen.append(msg.payload.decode())

# Maak een nieuwe MQTT-client instantie aan met de nieuwste API-versie (VERSION2)
client = mqtt.Client(mqtt.CallbackAPIVersion.VERSION2)

# Koppel onze eigen bij_bericht-functie aan het event 'on_message' van de client
client.on_message = bij_bericht

# Maak een netwerkverbinding met de MQTT-broker
client.connect(BROKER, POORT, keepalive=60)

# Start een achtergrond-thread (loop) die netwerkverkeer en inkomende berichten verwerkt
client.loop_start()

# Abonneer de client op ons unieke test-topic
client.subscribe(TOPIC)

# Wacht 1 seconde zodat de broker de abonnering verwerkt heeft
time.sleep(1)

# Publiceer een testbericht op het unieke topic
client.publish(TOPIC, "hallo vanuit mijn notebook")

# Wacht maximaal 5 seconden (50 stappen van 0.1s) totdat het bericht via de broker is ontvangen
for _ in range(50):
    if ontvangen:
        break                   # Bericht is binnen, verlaat de wacht-loop
    time.sleep(0.1)

# Stop de achtergrond-thread en verbreek de verbinding op een schone manier
client.loop_stop()
client.disconnect()

# Toon de resultaten in het uitvoerblok
print("Getest topic:", TOPIC)
print("Ontvangen  :", ontvangen or "NIETS — zie de netwerktip hieronder")
````

In dat geval kan MQTT verzonden worden via **WebSockets** met **TLS-versleuteling** op poort `8884`. Dit omzeilt firewalls omdat het verkeer er uitziet als beveiligd webverkeer (HTTPS). 

Mocht poort 1883 geblokkeerd zijn, vervang dan in de codecel hierboven de initialisatie en verbinding door:

```python

client.tls_set()                            # Schakel SSL/TLS versleuteling in
```

## 4. Verbinding met een HTTP-API testen

Naast MQTT is **HTTP** (Hypertext Transfer Protocol) het meest gebruikt protocol op het internet. HTTP werkt volgens een **Request-Response (vraag-antwoord)** model:
1. De **client** (jouw notebook of een app) verstuurt een expliciete vraag (**HTTP Request**) naar een webserver.
2. De **server** verwerkt het verzoek en geeft een antwoord (**HTTP Response**) terug, meestal met een **statuscode** (zoals `200 OK`) en gegevens in **JSON-formaat**.

Een **API** (Application Programming Interface) is een digitale ingang van een webserver waarmee computerprogramma's rechtstreeks gegevens kunnen opvragen.

Onderstaande code vraagt de actuele weersomstandigheden in Gent op bij de gratis webdienst **Open-Meteo** via een zogenaamd HTTP `GET`-verzoek.

````python
import requests  # Importeer de HTTP-bibliotheek voor Python
import json      # Importeer de json-bibliotheek om JSON-structuren overzichtelijk af te drukken

# Het basis-webadres (endpoint) van de Open-Meteo weerdienst API
url = "https://api.open-meteo.com/v1/forecast"

# Een Python dictionary met zoekparameters die als achtervoegsel aan de URL worden toegevoegd
parameters = {
    "latitude": 51.05,            # Geografische breedtegraad van Gent
    "longitude": 3.72,            # Geografische lengtegraad van Gent
    "current": "temperature_2m",  # Vraag specifiek de actuele temperatuur op 2 meter hoogte op
}

# Verstuur het HTTP GET-verzoek naar de webserver met een time-out van 10 seconden
antwoord = requests.get(url, params=parameters, timeout=10)

# Controleer de HTTP-statuscode die de server teruggeeft (200 betekent dat het verzoek geslaagd is)
print("HTTP-statuscode:", antwoord.status_code, "(200 = OK)")
print()

# Converteer de ontvangen JSON-tekst uit de body van het antwoord automatisch naar een Python dictionary
gegevens = antwoord.json()

# 1. Druk EERST de VOLLEDIGE JSON-respons van de server af (met inspringen voor leesbaarheid)
print("=== 1. VOLLEDIGE JSON-RESPONS VAN DE SERVER ===")
print(json.dumps(gegevens, indent=2))
print()

# 2. Selecteer en toon DAARNA PAS de gewenste specifieke gegevens
print("=== 2. SELECTIE VAN DE GEWENSTE GEGEVENS ===")
huidig = gegevens["current"]

# Druk het tijdstip en de gemeten temperatuur af
print(f"Gent, {huidig['time']} UTC: {huidig['temperature_2m']} °C")
````

````python
# Uitgewerkt voorbeeld voor de locatie Kortrijk
import requests
import json  # Importeer json om gegevens structureel en overzichtelijk af te drukken

# Basis-URL van de API
url = "https://api.open-meteo.com/v1/forecast"

# Definieer de geografische coördinaten van Kortrijk
gevraagd_lat, gevraagd_lon = 50.828, 3.265

# Stel de zoekparameters samen voor een uitgebreide actuele weeraanvraag
parameters = {
    "latitude": gevraagd_lat,
    "longitude": gevraagd_lon,
    "current": ",".join([
        "temperature_2m",
        "relative_humidity_2m",
        "apparent_temperature",
        "is_day",
        "snowfall",
        "weather_code",
        "cloud_cover",
        "pressure_msl",
        "surface_pressure",
        "wind_speed_10m",
        "wind_direction_10m",
        "wind_gusts_10m",
    ]),
    "cell_selection": "nearest",  # Kies het dichtstbijzijnde roosterpunt in het weermodel
}

# Verstuur het HTTP-verzoek naar de server
antwoord = requests.get(url, params=parameters, timeout=10)
antwoord.raise_for_status()       # Werpt een foutmelding op als de HTTP-statuscode niet OK is

# Verwerk het antwoord
gegevens = antwoord.json()

# 1. Druk EERST de VOLLEDIGE JSON-respons af
print("=== 1. VOLLEDIGE JSON-RESPONS (ACTUEEL WEER) ===")
print(json.dumps(gegevens, indent=2))
print()

# 2. Selecteer en toon DAARNA PAS de gewenste specifieke gegevens
print("=== 2. SELECTIE VAN DE GEWENSTE GEGEVENS ===")
huidig, eenheden = gegevens["current"], gegevens["current_units"]

# Toon de locatie en actuele waarden met hun meeteenheid
print(f"Gevraagd:    {gevraagd_lat}, {gevraagd_lon}")
print(f"Roosterpunt: {gegevens['latitude']}, {gegevens['longitude']}")
print("Tijdstip:   ", huidig["time"])
for sleutel, waarde in huidig.items():
    if sleutel in ("time", "interval"):
        continue
    print(f"{sleutel:22s}: {waarde} {eenheden[sleutel]}")

# Vraag een tijdsreeks op per uur voor meerdere dagen
reeks = requests.get(url, params={
    "latitude": gevraagd_lat,
    "longitude": gevraagd_lon,
    "hourly": "temperature_2m,relative_humidity_2m",
    "past_days": 1,
    "forecast_days": 2,
    "timezone": "Europe/Brussels",
    "cell_selection": "nearest",
}, timeout=10)
reeks.raise_for_status()

# Haal het uuroverzicht uit de JSON-respons
uurlijks_json = reeks.json()

# 3. Druk EERST de VOLLEDIGE JSON-respons af voor de uurlijkse tijdsreeks
print("\n=== 3. VOLLEDIGE JSON-RESPONS (UURLIJKSE TIJDSREEKS) ===")
print(json.dumps(uurlijks_json, indent=2))
print()

# 4. Selecteer en toon DAARNA PAS de gewenste reekselementen
print("=== 4. SELECTIE UIT DE TIJDSREEKS ===")
uurlijks = uurlijks_json["hourly"]
eenheid = uurlijks_json["hourly_units"]
tijden = uurlijks["time"]

print(f"{len(tijden)} uurlijkse tijdstappen ontvangen van {tijden[0]} tot {tijden[-1]}:")
for i in (0, -1):  # Toon de allereerste en de allerlaatste meting in de tijdsreeks
    print(f"{tijden[i]}: {uurlijks['temperature_2m'][i]} {eenheid['temperature_2m']}, "
          f"{uurlijks['relative_humidity_2m'][i]} {eenheid['relative_humidity_2m']}")
````

## 5. Samenvatting: Vergelijking van communicatiemodellen

In dit eerste notebook heb je de twee fundamentele communicatiemodellen van het Internet of Things uitgetest:

| Eigenschap | MQTT (Message Queuing Telemetry Transport) | HTTP (Hypertext Transfer Protocol) |
|---|---|---|
| **Communicatiemodel** | **Publish-Subscribe (Push)**: De zender verstuurt data naar de broker, die het direct 'duwt' naar ontvangers. | **Request-Response (Pull)**: De client vraagt expliciet data op en de server geeft één antwoord. |
| **Verbindingssnelheid** | Eén permanente TCP-verbinding die open blijft. Extreem snel met lage vertraging (*latency*). | Voor elke interactie wordt een nieuwe verbinding opgebouwd en afgebroken. |
| **Overhead** | Minimale netwerk-overhead (2 bytes per pakket). Ideaal voor batterijgevoede IoT-sensoren. | Grotere overhead door uitgebreide tekstuele HTTP-headers. |
| **Toepassing** | Live sensordatastromen, alarmen, directe aansturing van apparaten. | Opvragen van historische data, instellingen wijzigen, web-API's en dashboards. |
